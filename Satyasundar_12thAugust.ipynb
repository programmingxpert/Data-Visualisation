```python
import pandas as pd
import numpy as np
df = pd.read_csv("Data Cleaning 12.08.csv")
df.head()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
# Fix missing mass by combining mass and mass_sini
df['mass'] = df['mass'].fillna(df['mass_sini'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# drop rows that are still missing in mass column
df = df.dropna(subset=['mass'])
```

```python
df.isnull().sum()
```

```python
# Fix missing errors by fixing the empty values with 0 so that the calculations won't break
error_cols = ['mass_error_min', 'mass_error_max', 'mass_sini_error_min', 'mass_sini_error_max', 'radius_error_min']
for col in error_cols:
    df[col] = df[col].fillna(0)
```

```python
df.isnull().sum()
```

```python
# Fix missing radius
df['radius'] = df['radius'].fillna(df['radius'].median())
```

```python
df.isnull().sum()
```

```python
# drop the redundant mass_sini column
df = df.drop(columns=['mass_sini'])
```

```python
df.to_csv("Cleaned_Exoplanet_Data_Final.csv", index=False)
```

```python
df.describe()
```

```python
# Drops non-numeric columns like name automatically
df.corr(numeric_only=True)
```

```python
df['planet_status'].value_counts()
```

```python
# fix infinite values by replacing inf with NaN, then filling them with 0
df = df.replace([np.inf, -np.inf], np.nan)
df['mass_error_max'] = df['mass_error_max'].fillna(0)
```

```python
# fix negative values by converting negative error values to 0
df['mass_error_max'] = df['mass_error_max'].clip(lower=0)
```

```python
# drop planet_status because every row is 'Confirmed'
df = df.drop(columns=['planet_status'], errors='ignore')
```

```python
# save final dataset
df.to_csv("Cleaned_Exoplanet_Data_Final.csv", index=False)
```
